```python
# ==========================================
# Employee Attrition Analysis Script
# Task: Predict employee churn using ML models
# Name: Muhammad Rizwan
# Roll No.:097
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ML Libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# ------------------------------------------
# Step 1: Load and Inspect Dataset
# ------------------------------------------
print("--> Loading the IBM HR Attrition dataset...")
file_path = "WA_Fn-UseC_-HR-Employee-Attrition.csv"
data = pd.read_csv(file_path)

print(f"Dataset successfully loaded with {data.shape[0]} rows and {data.shape[1]} columns.\n")

# Check missing values just to be sure
missing_val_count = data.isnull().sum().sum()
print(f"Total missing values in dataset: {missing_val_count}\n")


# ------------------------------------------
# Step 2: Data Cleaning & Preprocessing
# ------------------------------------------
print("--> Cleaning unnecessary columns and encoding categorical variables...")

# Removing columns that don't add predictive value
drop_cols = ["EmployeeCount", "EmployeeNumber", "Over18", "StandardHours"]
data_cleaned = data.drop(columns=drop_cols, errors='ignore')

# Mapping Target Variable: Attrition ('Yes' -> 1, 'No' -> 0)
data_cleaned['Attrition'] = data_cleaned['Attrition'].map({'Yes': 1, 'No': 0})

# Separating features (X) and target (y)
X = data_cleaned.drop(columns=['Attrition'])
y = data_cleaned['Attrition']

# Converting categorical text features )
X_encoded = pd.get_dummies(X, drop_first=True)

print(f"Total features after One-Hot Encoding: {X_encoded.shape[1]}\n")


# Step 3: Train-Test Split & Scaling
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardizing numerical features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


# Step 4: Model Training & Comparisons
clf_models = {
    "Logistic Regression": LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(class_weight='balanced', random_state=42),
    "Random Forest": RandomForestClassifier(class_weight='balanced', n_estimators=100, random_state=42)
}

results_summary = {}

print("--> Training models and generating performance metrics...\n")

for model_name, model_obj in clf_models.items():
    # Model Training
    model_obj.fit(X_train_scaled, y_train)

    # Model Prediction
    y_predictions = model_obj.predict(X_test_scaled)

    # Computing Metrics
    acc = accuracy_score(y_test, y_predictions)
    prec = precision_score(y_test, y_predictions)
    rec = recall_score(y_test, y_predictions)
    f1 = f1_score(y_test, y_predictions)
    cm = confusion_matrix(y_test, y_predictions)

    # Storing results
    results_summary[model_name] = {
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1 Score': f1,
        'Confusion Matrix': cm
    }

    # Print formatted output
    print(f"================ {model_name} ================")
    print(f"Accuracy : {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall   : {rec:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print("Confusion Matrix:")
    print(cm)
    print("\n")


# Step 5: Visualizing Confusion Matrices
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, (name, metrics) in enumerate(results_summary.items()):
    sns.heatmap(
        metrics['Confusion Matrix'],
        annot=True,
        fmt='d',
        cmap='Blues',
        ax=axes[i],
        cbar=False
    )
    axes[i].set_title(f"{name}\nF1 Score: {metrics['F1 Score']:.2f}")
    axes[i].set_xlabel("Predicted Label")
    axes[i].set_ylabel("Actual Label")

plt.suptitle("Confusion Matrix Comparison Across Models", fontsize=14)
plt.tight_layout()
plt.show()


# Step 6: Feature Importance Analysis (Random Forest)

rf_classifier = clf_models["Random Forest"]

# Extracting feature importances
importance_vals = rf_classifier.feature_importances_
feature_names = X_encoded.columns

# Creating a DataFrame for top features
feat_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance_vals
}).sort_values(by='Importance', ascending=False)

# Plotting Top 12 Features
plt.figure(figsize=(10, 6))
sns.barplot(
    x='Importance',
    y='Feature',
    data=feat_df.head(12),
    palette='Blues_r'
)

plt.title("Top 12 Most Important Features for Predicting Attrition")
plt.xlabel("Importance Score")
plt.ylabel("Feature Name")
plt.tight_layout()
plt.show()
```
